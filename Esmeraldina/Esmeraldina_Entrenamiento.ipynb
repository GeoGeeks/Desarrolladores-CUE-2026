```python
archivo = r"C:\Users\perey\OneDrive - Esri Colombia, Ecuador y Panamá\Esmeraldina\Fase IV. Machine Learning\Esmeraldina-ArcGISPro\CSV'S\Dataset_Simulacion_Esmeraldina_FaseIII_Calificado.csv"

df = pd.read_csv(archivo, sep=";")

print("CSV cargado correctamente.")
print("Número de muestras:", df.shape[0])
print("Número de columnas:", df.shape[1])

df.head()
```

```python
print("COLUMNAS DEL DATASET:\n")

for i, columna in enumerate(df.columns, start=1):
    print(i, "-", columna)
```

```python
# Variables explicativas
columnas_X = [
    "LITOLOGIA",
    "DIST_FALLA_m",
    "DIST_INTERSECCION_m",
    "GUIA_PIRITA",
    "Na_pct",
    "K_pct",
    "CR_ppm",
    "V_ppm",
    "K_Na",
    "Be_ppm"
]

# X = información que utilizará el modelo
X = df[columnas_X].copy()

# y = valor que queremos que el modelo aprenda a predecir
y = df["IPE"].copy()

print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)

print("\nPrimeras filas de X:")
display(X.head())

print("\nPrimeros valores de y:")
display(y.head())
```

```python
print("TIPOS DE DATOS DE X:\n")
print(X.dtypes)

print("\nTIPO DE DATO DE y:")
print(y.dtype)
```

```python
print("VALORES ÚNICOS DE GUIA_PIRITA:")
print(X["GUIA_PIRITA"].unique())

print("\nVALORES ÚNICOS DE LITOLOGIA:")
print(X["LITOLOGIA"].unique())
```

```python
import sklearn

print("Scikit-learn instalado correctamente.")
print("Versión:", sklearn.__version__)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=2026
)

print("DATOS DE ENTRENAMIENTO")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nDATOS DE PRUEBA")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# Variable categórica sin orden
variable_litologia = ["LITOLOGIA"]

# Variable categórica con orden
variable_pirita = ["GUIA_PIRITA"]

# Variables que ya son numéricas
variables_numericas = [
    "DIST_FALLA_m",
    "DIST_INTERSECCION_m",
    "Na_pct",
    "K_pct",
    "CR_ppm",
    "V_ppm",
    "K_Na",
    "Be_ppm"
]

# Codificación ordenada de la guía de pirita
codificador_pirita = OrdinalEncoder(
    categories=[["Ausente", "Baja", "Media", "Alta"]]
)

# Preprocesador
preprocesador = ColumnTransformer(
    transformers=[
        (
            "litologia",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            variable_litologia
        ),
        (
            "pirita",
            codificador_pirita,
            variable_pirita
        ),
        (
            "numericas",
            "passthrough",
            variables_numericas
        )
    ]
)

print("Preprocesador creado correctamente.")
```

```python
X_train_procesado = preprocesador.fit_transform(X_train)

X_test_procesado = preprocesador.transform(X_test)

print("X_train original:", X_train.shape)
print("X_train procesado:", X_train_procesado.shape)

print("\nX_test original:", X_test.shape)
print("X_test procesado:", X_test_procesado.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

# Crear el modelo
modelo_rf = RandomForestRegressor(
    n_estimators=300,
    random_state=2026,
    n_jobs=-1
)

# Entrenar el modelo
modelo_rf.fit(
    X_train_procesado,
    y_train
)

print("Random Forest entrenado correctamente.")
print("Número de árboles:", modelo_rf.n_estimators)
print("Muestras utilizadas para entrenamiento:", X_train_procesado.shape[0])
print("Variables utilizadas por el modelo:", X_train_procesado.shape[1])
```

```python
y_pred = modelo_rf.predict(X_test_procesado)

print("Predicciones realizadas correctamente.")
print("Número de predicciones:", len(y_pred))

print("\nPrimeros 10 IPE predichos:")
print(y_pred[:10])
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

# Calcular métricas
mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

print("EVALUACIÓN DEL MODELO")
print("---------------------")
print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 4))
```

```python
comparacion = pd.DataFrame({
    "IPE_REAL": y_test.values,
    "IPE_PREDICHO": y_pred
})

comparacion["ERROR"] = (
    comparacion["IPE_REAL"] -
    comparacion["IPE_PREDICHO"]
)

comparacion["ERROR_ABS"] = comparacion["ERROR"].abs()

comparacion.head(10)
```

```python
# Obtener nombres de las variables después del preprocesamiento
nombres_variables = preprocesador.get_feature_names_out()

print("Número de variables:", len(nombres_variables))

for variable in nombres_variables:
    print(variable)
```

```python
# Obtener importancia de cada variable
importancias = modelo_rf.feature_importances_

importancia_df = pd.DataFrame({
    "VARIABLE": nombres_variables,
    "IMPORTANCIA": importancias
})

# Ordenar de mayor a menor
importancia_df = importancia_df.sort_values(
    by="IMPORTANCIA",
    ascending=False
)

# Convertir a porcentaje
importancia_df["IMPORTANCIA_PCT"] = (
    importancia_df["IMPORTANCIA"] * 100
)

importancia_df.head(15)
```

```python
# Crear copia para agrupar importancias
importancia_agrupada = importancia_df.copy()

def obtener_parametro(nombre):
    
    if nombre.startswith("litologia__"):
        return "LITOLOGIA"
    
    elif nombre.startswith("pirita__"):
        return "GUIA_PIRITA"
    
    elif nombre.startswith("numericas__"):
        return nombre.replace("numericas__", "")
    
    else:
        return nombre

importancia_agrupada["PARAMETRO"] = (
    importancia_agrupada["VARIABLE"].apply(obtener_parametro)
)

# Sumar importancias de categorías que pertenecen al mismo parámetro
importancia_parametros = (
    importancia_agrupada
    .groupby("PARAMETRO", as_index=False)["IMPORTANCIA"]
    .sum()
)

# Convertir a porcentaje
importancia_parametros["IMPORTANCIA_PCT"] = (
    importancia_parametros["IMPORTANCIA"] * 100
)

# Ordenar
importancia_parametros = importancia_parametros.sort_values(
    by="IMPORTANCIA_PCT",
    ascending=False
)

importancia_parametros
```

```python
# Recuperar las 60 muestras originales utilizadas como prueba
resultados_60 = df.loc[X_test.index].copy()

# Agregar el IPE calculado por el modelo
resultados_60["IPE_ML"] = y_pred

# Calcular diferencia entre IPE original e IPE del modelo
resultados_60["ERROR_ML"] = (
    resultados_60["IPE"] - resultados_60["IPE_ML"]
)

resultados_60["ERROR_ABS_ML"] = (
    resultados_60["ERROR_ML"].abs()
)

print("Número de muestras:", len(resultados_60))

resultados_60[
    [
        "ID_MUESTRA",
        "LATITUD",
        "LONGITUD",
        "IPE",
        "IPE_ML",
        "ERROR_ML",
        "ERROR_ABS_ML"
    ]
].head(10)
```

```python
from sklearn.pipeline import Pipeline

esmeraldina = Pipeline(
    steps=[
        ("preprocesador", preprocesador),
        ("random_forest", modelo_rf)
    ]
)

print("Esmeraldina V1 preparada correctamente.")
```

```python
# Predicción usando Esmeraldina completa
prueba_esmeraldina = esmeraldina.predict(X_test)

# Comprobar que produce los mismos resultados
diferencia_maxima = abs(prueba_esmeraldina - y_pred).max()

print("Diferencia máxima:", diferencia_maxima)

if diferencia_maxima < 0.000001:
    print("Esmeraldina funciona correctamente.")
else:
    print("Revisar el modelo antes de guardarlo.")
```
